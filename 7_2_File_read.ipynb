```python
#read function.
file = open("./sample_data/sample.txt", "r")
content = file.read()
print(type(content))
print(content)


#readlines function: newline thakle '\n' show kore.
file = open("./sample_data/sample.txt", "r")
content = file.readlines()
print(type(content))
print(content)

#file closing with close() function:
#file jate leackage na hoy and vlo vabe save hoy ejonno file close ensure kora dorkar.
file.close()
print(file.closed)
```

**Closing with "with"**

```python
#'with' diye file nile alada kore r close function use korte hoy na.
with open("./sample_data/sample.txt", "r") as files: #files is a object here
  content = files.readlines()
  print(content)
print(files.closed)
```

**Showing result Using for loop**

```python
#for loop er maddhome line by line print kora jay.
with open("./sample_data/sample.txt", "r") as files:
  for line in files:
    #l = line.strip() #strip() nile 'end' likhte hoy na r.
    print(line, end=" ")  #extra faka akta line reduce korar jonno 'end' newa hoice.
```
