```python
#function return type:
def give_prediction():
  return 10 #int
print(give_prediction())

#function return type:
def give_prediction_1():
  return 10.6 #float
print(give_prediction_1())

#function return type:
def give_prediction_2():
  return "Rasel"  #string
print(give_prediction_2())

#function return type:
def give_prediction_3():
  return 10,14  #comma diye likhle tuple hoy
print("Tuple : ", give_prediction_3())


#Unpacking
def give_prediction_4():
  return 13,64  #comma diye likhle tuple hoy
print("Tuple : ", give_prediction_4())
x,y = give_prediction_4() #x,y 2ta sadhin value akhon.
print("Unpacked value : ", x, y)
```

**List Return**

```python
#List return (packing/unpacking)
def give_prediction_5():
  return [10, 20 ,30, [1, 2, 3]]  #list akare pathalei list hbe.
print("List : ", give_prediction_5())

#packing / unpacking list
def give_prediction_6():
  a = [10,20,30]
  b = [12,22,33]
  return a,b  #tuple akare list 2ta k pathabe.
a, b = give_prediction_6()
print("1st list : ", a)
print("2nd list : ", b)
```

**Set return**

```python
#Set returning
def give_prediction_7():
  a = {1, 2, 3, 6}
  return a
x = give_prediction_7()
print(type(x))
print(x)

for item in x:  #value printing
  print(item)
```

**Dictionary return**

```python
#Dictionary return
def give_prediction_8():
  a = {"name":"Rasel", "addr":"nil", "cell":"734"}, {"name":"Imran", "addr":"domar", "cell":"744"}
  return a
x,y = give_prediction_8()
print(x)
print(y)

print("\n")

print("1st dict : ")
for i in x.items():
  print(i)

print("\n")

print("2nd dict : ")
for j in y.items():
  print(j)
```
