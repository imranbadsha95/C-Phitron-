```python
from functools import reduce #(eta built in na jonne import korte hbe surute)

lst = [1,2,3,4,5,6]
summation = reduce(lambda x,y: x+y, lst)
print(summation)

#suru theke prottek 2 number niye segulo gun kore akta result ber kora hocche.
#ei result k abar next value diye gun kora hocche. and j resul aslo setake abar mone rakha hocce.
#then ei result er sathe abaro next value diye gun. evbe sese giye purotar akta result asbe.
#evabe final result ta return kora hocche.
#jog er belay jog, biyog hole biyog. gun bolle gun and vag bolle vag kora hcce.
#onno operation kora hole setar result sevabei asbe.

lst2 = [1,2,3,4,5,6]
summation = reduce(lambda x,y: x*y, lst2)
print(summation)

lst3 = [1,2,3,4,5,6]
max_val = reduce(lambda a,b: a if a>b else b, lst3) #a boro hole b choto hle.
print(max_val)
```

```python
#to find largest value.
numbers = [10, 5, 25, 8, 30, 15]

max_value = reduce(lambda x, y: x if x > y else y, numbers)
print(max_value)  # Output: 30


numbers = [1, 2, 3, 4]
# Initializer 10 দিয়ে শুরু
result = reduce(lambda x, y: x + y, numbers, 10)
print(result)  # Output: 20 (10 + 1 + 2 + 3 + 4)
```

```python
from functools import reduce
sales = [
    {'product': 'A', 'amount': 100},
    {'product': 'B', 'amount': 200},
    {'product': 'A', 'amount': 150},
    {'product': 'C', 'amount': 300}
]

# Product A er total sales
total_sales_a = reduce(
    lambda total, sale: total + sale['amount'] if sale['product'] == 'A' else total,
    sales,
    0  # initial value
)
print(f"Total sales for product A: {total_sales_a}")
# Output: Total sales for product A: 250
```
