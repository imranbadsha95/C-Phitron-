```python
#map : map(function, iterable)  #function ta muloto object akare dite hoy. without parenthesis
lst = [10, 20, 30, 40]

def sq(x):
  return x**2
lst11 = list(map(sq,lst)) #sq function but object akare dite hoy ekhane.
print(lst11)
```

### **String**

```python
string_1 = "Hello, how are you?"
string_2 = list(map(str.upper, string_1))
print(string_2)

#joining string
string_2 = "".join(string_2)
print(string_2)
```

**Using lambda function**

```python
list_1 = [1, 2, 3, 4, 5, 6]
result = list(map(lambda x: x**2 if x%2==0 else x, list_1))
print(result)
```
