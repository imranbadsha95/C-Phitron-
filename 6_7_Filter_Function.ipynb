```python
#filter function
num = [x for x in range(100)]
print(num)

even = list(filter(lambda x : x %2==0, num))
print(even)

fifty_upper = list(filter(lambda x : x > 50, num))
print(fifty_upper)

#none data
data = [0, 1, '', None, 'Hello', [], [1,2], None, True, False,None]
cleaned_data = list(filter(None, data))
print(cleaned_data)
```

**String**

```python
stringz = "Hello there, how are you?"
vowel = list(filter(lambda x : x in 'aeiou', stringz))
print(vowel)


fruits = ["app", "banana", 'dab', "mango", 'pomegranate']
filtering = list(filter(lambda x : len(x)>4, fruits))
print(filtering)
```

```python
# Email validation
emails = ["userexample.com","invalid-email","test@gmail.com","another@domain.org","wrong-format"]

valid_emails = filter(lambda email: '@' in email and '.' in email, emails)
print(list(valid_emails))
```

```python
students = [
  {'name': 'Alice', 'age': 20, 'grade': 'A'},
  {'name': 'Bob', 'age': 22, 'grade': 'B'},
  {'name': 'Charlie', 'age': 19, 'grade': 'A'},
  {'name': 'David', 'age': 21, 'grade': 'C'}
]

# 'A' Grade and age 21
top_young_students = filter(lambda s: s['grade'] == 'A' and s['age'] < 21, students)
print(list(top_young_students))
```
