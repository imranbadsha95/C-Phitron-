```python
#There are 2 kind of arguments
#1. Positional argument
#2. Keyword argument

#function declarations.
def greet(user ="guest"): #'guest' is default parameter.
  print(f"Hello {user}")

greet("Imran")
```

```python
#without using returning method.
def square_addition(a,b):
  a = a**2
  b = b**2
  print(a+b)
square_addition(2,3)

#with using returning method, positional argumen:
#positional argument: j serial e parameter pathalam se serial e count hlo.
def square_addition_2(a,b):
  print(f"{a} {b}")
  a = a**2
  b = b**2
  return a+b
ans = 10 + square_addition_2(10,20) #return holo and + 10 holo
print(ans)

#keyword argument
def square_addition_3(a,b,c):
  print(f"{a} {b} {c}")
  a = a**2
  b = b**2
  return a+b+c
ans2 = square_addition_3(b=10,c=20, a=30)
print(ans2)
```

```python
#variable size argument (more than 3 parameter):
def square_addition_4(a,b,c):
  print(f"{a} {b} {c}")
  a = a**2
  b = b**2
  c = c**2
  return a+b+c
ans4 = square_addition_4(10,20,30)
print(ans4)
type(ans4)
```

```python
#variable size argument (more than 3 parameter):
def square_addition_4(*args):
  for i in args:
    print(i)
ans4 = square_addition_4(10,20,30, 40, 50, 60)
print(ans4)
```

```python
#opeation: more than 3 arguments in function
def square_addition_5(*args):
  sum = 0
  for i in args:
    i = i**2
    sum += i
  return sum

ans5 = square_addition_5(10,20,30, 40, 50)
print(ans5)
```

```python
#Keyword arguments: dictionary akare pass kore.
def f(**kwargs):  #key value pair = dictionary return kore.
  for key,val in kwargs.items():
    print(f"{key} : {val}")

f(name="Rasel", roll=20, marks=45.5)  #function
```
