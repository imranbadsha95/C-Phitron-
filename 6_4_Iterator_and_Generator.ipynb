**iterator**

```python
#iterator holo built in function
s = {10, 20, 30, 40}  #indexing maintain kore na. random value ase.
s_iter = iter(s)
#It cannot access items in a set by referring to an index, since sets are unordered the items has no index.

print(next(s_iter))
print(next(s_iter))
print(next(s_iter))
print(next(s_iter))
#print(next(s_iter))   #value er cheye besi hole error dey.

#using for loop: best way to access
print("\n")
for i in s: #set k loop diye iterate kora jay.
  print(i)
```

**Generator**

```python
lst = [x for x in range(500)]

def data_loader(chunk_size, lst):
  for i in range(0, len(lst), chunk_size):  #(start, end, step)
    yield lst[i:i+chunk_size] #5ta 5ta kore value niye kaj hcce.

x = data_loader(5, lst)

print(type(x))  #generator
print(x)  #x holo generator object
print(next(x))  #output: 0 to 4 #list akare val dicce.
print(next(x))  #output: 5 to 9
print(next(x))  #output: 10 to 14

#chunk_size: boro dataset theke part part akare value ney. purotai akbare ney na. limit dite hoy
#jeta yield keyword er maddhome korte hoy. Ex: 500 size theke 5 ta kore value niye dekha..
#boro data theke choto choto akare date niye dekha & analyz kora ba operation korar jonno ei system.
```
